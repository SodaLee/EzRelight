```python
from safetensors.torch import load_file, save_file
from models.unet import UNet2DConditionModel
import copy
unet = UNet2DConditionModel.from_pretrained(
        "stabilityai/stable-diffusion-xl-base-1.0", subfolder="unet", revision=None, variant='fp16', use_safetensors=True,
    )
unet_path = '/data2/lihaochen/EzRelight_chps/stage1/try_1115/checkpoints/final/unet_weight_increasements.safetensors'

unet_sd = load_file(unet_path)
orig_unet_sd = copy.deepcopy(unet.state_dict())
```

```python
imp_sd = {}

for k in orig_unet_sd.keys():
    if k in unet_sd.keys():
        imp_sd[k] = unet_sd[k] + orig_unet_sd[k]
    else:
        imp_sd[k] = orig_unet_sd[k]

unet.load_state_dict(imp_sd)
```
